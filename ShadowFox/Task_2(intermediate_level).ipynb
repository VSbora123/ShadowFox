```python
# Q1. Web Scraping..
import requests
from bs4 import BeautifulSoup
```

```python
page_count = 1
while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)

    soup = BeautifulSoup(res.text, "lxml")
    quotes = soup.select("div.quote")

    if not quotes:
        print("no valid pages anymore..")
        break

    with open(f"scraped_data/quotes{page_count}.html", "w", encoding = "utf-8") as f:
        f.write(res.text)
        print(f"downloaded from page {page_count}")

    page_count += 1
```

```python
for i in range(1, 11):
    with open(f"scraped_data/quotes{i}.html", "r", encoding = "utf-8") as f:
        html_content = f.read()

    soup = BeautifulSoup(html_content, "lxml")

    all_quotes = soup.select("div.quote")
    life_quotes = []

    for q in all_quotes:
        all_tags = []

        for tag in q.select(".tags .tag"):
            all_tags.append(tag.get_text())

        if "inspirational" in all_tags:
            text = q.select_one("span.text").get_text()
            author = q.select_one("small.author").get_text()
            life_quotes.append([text, author])

    for quote in life_quotes:
        print(f"\n{quote[0]} — {quote[1]}")
```

```python
import random

def play_hangman():
    data = {"apple": "A red fruit", "grapes": "A green fruit", "python": "A programming language"}
    
    word = random.choice(list(data.keys()))
    hint = data[word]

    guessed_letters = []
    lives = 6

    print("-----Welcome to Hangman-----")

    while lives > 0:
        display = ""
        for char in word:
            if char in guessed_letters:
                display += char
            else:
                display += "_"

        print(f"\nword: {display}")
        print(f"Lives: {lives}")

        if "_" not in display:
            print("You won!")
            return

        guess = input("Guess a letter( or type 'help' for hint): ").lower()

        if guess == "help":
            print(f"HINT: {hint}")
            continue

        if guess in guessed_letters:
            print("You already tried that")
            continue

        guessed_letters.append(guess)

        if guess in word:
            print("Good guess!")
        else:
            print("Not in word.")
            lives -= 1
    if lives == 0:
        print(f"\nGame Over! The word was: {word}")
        
play_hangman()
```
